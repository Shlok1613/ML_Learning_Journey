```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('darkgrid')
%matplotlib inline
```

```python
iris_df = sns.load_dataset('iris')
```

```python
iris_df
```

```python
sns.scatterplot(data=iris_df, x="sepal_length", y="petal_length")
plt.show()
```

```python
sns.scatterplot(data=iris_df, x="sepal_width", y="petal_width")
plt.show()
```

```python
numeric_cols=['petal_width', 'petal_length', 'sepal_length', 'sepal_width']
```

```python
X = iris_df[numeric_cols]
```

```python
X
```

<h1>K Means Clustering</h1>

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3, random_state=42)
```

```python
model.fit(X)
```

```python
model.cluster_centers_
```

```python
X
```

```python
preds = model.predict(X)
preds
```

```python
sns.scatterplot(data=X, x='sepal_length', y='petal_length', hue=preds);
centers_x, centers_y = model.cluster_centers_[:,2], model.cluster_centers_[:,1]
plt.plot(centers_x, centers_y, 'xb')
plt.show()
```

```python
model.inertia_
```

```python
model = KMeans(n_clusters=6, random_state=42)
```

```python
model.fit(X)
```

```python
model.cluster_centers_
```

```python
preds = model.predict(X)
preds
```

```python
sns.scatterplot(data=X, x='sepal_length', y='petal_length', hue=preds);
plt.show()
```

```python
model.inertia_
```

```python
options = range(2,11)
inertias = []

for n_clusters in options:
    model = KMeans(n_clusters, random_state=42).fit(X)
    inertias.append(model.inertia_)
    
plt.title("No. of clusters vs. Inertia")
plt.plot(options, inertias, '-o')
plt.xlabel('No. of clusters (K)')
plt.ylabel('Inertia')
plt.show()
```

<h1>DBSCAN</h1>

```python
?DBSCAN
```

```python
model = DBSCAN(eps=1.1, min_samples=4)
```

```python
model.fit(X)
```

```python
model.labels_
```

```python
dir(model)
```

```python
model.core_sample_indices_
```

```python
sns.scatterplot(data=X, x='sepal_length', y='petal_length', hue=model.labels_)
plt.show()
```

```python
def testing(x, y):
    model = DBSCAN(eps=x, min_samples=y).fit(X)
    sns.scatterplot(data=X, x='sepal_length', y='petal_length', hue=model.labels_)
    plt.show()
```

```python
testing(1, 3)
```

```python
testing(1.5, 3)
```

```python
testing(0.5, 4)
```

<h1>PCA: Principle Component Analysis</h1>

```python
iris_df = sns.load_dataset('iris')
iris_df
```

```python
numerical_cols = ["sepal_length" , "sepal_width", "petal_length", "petal_width"]
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca.fit(iris_df[numeric_cols])
```

```python
transformed = pca.transform(iris_df[numeric_cols])
```

```python
sns.scatterplot(x=transformed[:,0], y=transformed[:,1], hue=iris_df['species']);
plt.show()
```

<h1>t-Distributed Stochastic Neighbor Embedding (t-SNE)</h1>

```python
from sklearn.manifold import TSNE
```

```python
tsne = TSNE(n_components=2)
```

```python
transformed = tsne.fit_transform(iris_df[numeric_cols])
```

```python
sns.scatterplot(x=transformed[:,0], y=transformed[:,1], hue=iris_df['species']);
```
